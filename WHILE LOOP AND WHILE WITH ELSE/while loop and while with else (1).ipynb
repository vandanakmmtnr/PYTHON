**Part A: While Loop**


1.Print the text "Day 1", "Day 2", ... up to "Day 7" using a while loop.

```python
day = 1
while day <= 7:
    print("Day", day)
    day += 1
```

2.The daily sales of a shop for 5 days are stored in a list:
Print each day's sales one by one using an index.



```python
sales = [1200, 1500, 900, 2000, 1800]
i = 0
while i < len(sales):
    print(sales[i])
    i += 1
```

3.Using the same sales list, find and print the total sales.


```python
sales = [1200, 1500, 900, 2000, 1800]
i = 0
total = 0
while i < len(sales):
    total += sales[i]
    i += 1
print("Total sales:", total)
```

4.The marks of 5 students are given below. Calculate and print the average marks.

```python
marks = [45, 78, 62, 90, 55]
i = 0
total = 0
while i < len(marks):
    total += marks[i]
    i += 1
average = total / len(marks)
print("Average marks:", average)
```

5.Using the same marks list, count how many students scored 50 or more (passed). Print the count.


```python
marks = [45, 78, 62, 90, 55]
i = 0
count = 0
while i < len(marks):
    if marks[i] >= 50:
        count += 1
    i += 1
print("Students passed:", count)
```

6.A product costs ₹25. Take the quantity from the user and print the cost for 1 item, 2 items, 3 items ... up to the given quantity. (Example for quantity 3: 25, 50, 75)


```python
quantity = int(input("Enter quantity: "))
price = 25
i = 1
while i <= quantity:
    print(price * i)
    i += 1
```

7.Find the highest value in the following list using a while loop (do not use max()):


```python
temperatures = [31, 34, 29, 36, 33]
highest = temperatures[0]
i = 1
while i < len(temperatures):
    if temperatures[i] > highest:
        highest = temperatures[i]
    i += 1
print("Highest temperature:", highest)
```

**Part B: While with else**


8.Print the month numbers 1 to 3 using a while loop. After the loop finishes, print "Report Generated" using an else block.


```python
month = 1
while month <= 3:
    print(month)
    month += 1
else:
    print("Report Generated")
```

9.A list of customer IDs is given:
ids = [101, 102, 103, 104]
Print each ID using a while loop, and then print "All customers listed" using an else block.


```python
ids = [101, 102, 103, 104]
i = 0
while i < len(ids):
    print(ids[i])
    i += 1
else:
    print("All customers listed")
```
