10. A taxi charges ₹60 for the first 2 km and ₹15 for every extra km. If the ride is between 10 PM and
5 AM, a night charge of 25% is added to the total fare. Take the distance and the hour (0 to 23)
and print the final fare.

```python
d = float(input("Distance (km): "))
hour = int(input("Hour (0-23): "))
fare = 60 if d <= 2 else 60 + (d - 2) * 15
if hour >= 22 or hour < 5:
    fare = fare * 1.25
print("Final fare: ₹", fare)
```

15.Write a short note with an example: why does the comparison 0.1 + 0.2 == 0.3 print False in Python? How can you check the condition correctly using abs() and a small difference?


```python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
```

```python
a = 0.1 + 0.2
b = 0.3

if abs(a - b) < 1e-9:
    print("Equal")
else:
    print("Not equal")
```

Computers store floats in binary, and values like 0.1 and 0.2 can't be represented exactly. They are stored as close approximations, so the sum has a tiny error.
So instead of checking for exact equality, check whether the difference is smaller than a tiny value like 1e-9.
