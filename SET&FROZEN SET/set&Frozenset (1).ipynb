c) Explain why a frozenset can be used as a dictionary key or inside another set, but a normal set cannot.


 Set elements and dictionary keys must be hashable. A hash must never change while the object is in the collection. A normal set is mutable, so its contents (and hash) could change, and Python refuses to hash it. A frozenset can never change, so it is safe to hash.

D) Give one real-life use case for a set and one for a frozenset

 A set suits tracking unique visitors to a website, because you keep adding to it. A frozenset suits a fixed group of permissions or tags used as a dictionary key, such as frozenset({"read", "write"}).

*Question 8: Unique Pairs Using Frozenset*


```python
pairs = [
    ("Asha", "Bala"),
    ("Bala", "Asha"),
    ("Chitra", "Dinesh"),
    ("Asha", "Chitra"),
    ("Dinesh", "Chitra"),
    ("Asha", "Bala")
]
```

a) If you put these tuples directly into a set, how many unique pairs do you get? Is that correct for this problem? Why not?

```python
set(pairs)   
```

 get 5, which is wrong for this problem. Tuples are ordered, so ("Asha","Bala") and ("Bala","Asha") count as different, and likewise for the Chitra/Dinesh pair.

b) Convert each pair into a frozenset and store them in a set. How many unique conversations are there now?

```python
unique = {frozenset(p) for p in pairs} 
print(unique)
```

 A frozenset ignores order, so there are 3 unique conversations.

c) Why can't you use a normal set instead of a frozenset inside the outer set here?


 A normal set is unhashable, so it can't be an element of another set or a dictionary key. A frozenset is hashable, so it can.

d) Create a dictionary that counts how many times each conversation appears, using the frozenset as the key.


```python
count = {}
for pair in pairs:
    conversation = frozenset(pair)
    count[conversation] = count.get(conversation, 0)+1
print(count)
```
