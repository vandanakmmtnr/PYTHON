Integer is a datatype used to specifies the numbers

```python
Salary = 20000.20
print(type(Salary))
```

Float is the datatype used to specifies the fractional numbers.

```python
Mark = True
print(type(Mark))
```

Boolean is the datatype used to printed true or false values only

```python
fruits = ["Apple","Banana","Orange"]
print("the datatype of fruits :",(type(fruits)))
```

List is the collection of multiple items stored in a single variable 

```python
coordinates = (10,20)
print(type(coordinates))
```

A tuple is an ordered collection of values that is immutable , meaning its values cannot be changed after it is created

```python
Student = {"name": "Vandana","age": "21"}
print("The datatype of Student:",(type(Student)))
```

A dictionary is a collection of key_value pairs used to store data in Python

```python
Numbers = {1,2,3,4}
print("The datatype of Numbers:",(type(Numbers)))
```

A set is an unordered collection of unique elements in Python.It does not allow duplicate value.
