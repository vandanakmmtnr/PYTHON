**Part B: Find and Fix the Error**

**B1**

```python
fruits = ["Apple","Mango"]
fruits.extend(["Orange","Grapes"])
print(fruits)
```

mistake: append() takes only one argument, so it can't add two items at once. This gives a TypeError.

**B2**

```python
numbers = [10, 20, 30]
numbers.pop(2)
print(numbers)
```

Mistake: remove() deletes by value, not by index. remove(2) looks for the value 2, which isn't in the list, so it raises a ValueError. To delete the item at index 2, use pop().

**B3**

```python
numbers = [30, 10, 20]
numbers.sort()
print(numbers)
```

Mistake: sort() sorts the list in place and returns None. Assigning its result back to numbers overwrites the list with None.

**B4**

```python
fruits = ["Apple", "Orange"]
print(fruits[1])
```

Mistake: The list has only two items, so the valid indexes are 0 and 1. fruits[2] is out of range and raises an IndexError.

**B5**

```python
numbers = [1, 2, 3]
numbers.append(4)
print(numbers)
```

Mistake: Lists don't have an add() method (that belongs to sets). This raises an AttributeError. To add an item to the end of a list, use append().

**D2**

```python
marks = [72, 85, 90, 65, 88, 76, 95, 58]

marks.sort(reverse=True)             
print(marks)

print("Top 3:", marks[:3])
print("Bottom 3:", marks[-3:])

total = sum(marks)
average = total / len(marks)
print("Total:", total)
print("Average:", average)
```

**D3**

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12000, 15000, 9000, 18000, 21000, 17000]
print("Total:",sum(sales)) #sum()= total

average = sum(sales) / len(sales) #len = function to count the length
print("Average:", round(average, 2)) #round = to round the value in fixed decimal place
```

```python
high_sale = max(sales)
print(high_sale)
print("Best month:",months[sales.index(high_sale)],"(",high_sale,")")
```

```python
count = 0
for sale in sales: 
    if sale >= average:
        count = count+1
print("Months above average:",count)
```
