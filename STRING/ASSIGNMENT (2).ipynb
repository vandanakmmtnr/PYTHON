**SECTION C**

```python
data = " rahul kumar,data analytics,rahul.kumar@gmail.com,9876543210 "

parts = data.strip().split(",")

name   = parts[0].strip().title()
course = parts[1].strip().title()
email  = parts[2].strip().lower()
phone  = parts[3].strip()

username = email[:email.find("@")]
masked   = phone[:2] + "*" * (len(phone) - 4) + phone[-2:]
words    = name.split()
initials = words[0][0] + "." + words[-1][0]
code     = course[:3].upper() + "-2026"

print("========== STUDENT ID CARD ==========")
print(f"Name        : {name}")
print(f"Course      : {course}")
print(f"Email       : {email}")
print(f"Username    : {username}")
print(f"Phone       : {masked}")
print(f"Initials    : {initials}")
print(f"Course Code : {code}")
print("=====================================")
```
