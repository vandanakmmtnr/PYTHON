**4**

```python
fruits = {"apple": 50, "banana": 30, "mango": 20}
item = fruits.popitem()
print(item)      # ('mango', 20)
print(fruits)    # {'apple': 50, 'banana': 30}
```

popitem() removes the last inserted pair.

**5**


```python
d = {"a": 1, "b": 2}
backup = d.copy()
d.clear()
print(d)        # {}
print(backup)   # {'a': 1, 'b': 2}
```

**6**

```python
d = {"a": 1, "b": 2, "c": 3}
d.pop("b")
d.popitem()
print(d)
```

{"a":1}
pop("b") removes b,
leaving {"a": 1, "c": 3}.
popitem() then removes the last pair, ("c", 3).

**3**

```python
dict1 = {"a": 1, "b": 2}
dict2 = {"b": 20, "c": 3}
dict1.update(dict2)
print(dict1)   # {'a': 1, 'b': 20, 'c': 3}
```

Why b changed: when a key exists in both dictionaries, update() overwrites the old value with the one from dict2. New keys (c) are simply added.

**4**

```python
a = {"x": 1}
b = a
b["x"] = 99
print(a)
```

b = a doesn't copy anything. Both names point to the same dictionary, so changing b changes a.

```python
a = {"x": 1}
b = a.copy()
b["x"] = 99
print(a)    # {'x': 1}
```

copy() creates a new, separate dictionary, so a is unaffected.

**Set C — Challenge**

**5. Shallow copy puzzle**

```python
a = {"name": "John", "marks": [80, 90]}
b = a.copy()
b["marks"].append(100)
b["name"] = "Sam"
print(a)
# {'name': 'John', 'marks': [80, 90, 100]}
```

name is a string. b["name"] = "Sam" rebinds b's key to a new string, so a keeps "John".
marks is a list shared by both dictionaries. append(100) modifies that one list in place, so a sees the change.
